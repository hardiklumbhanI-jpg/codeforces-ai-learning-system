# Recommendation and Similarity

Demonstrates the practice recommendation engine and the TF-IDF + cosine
similarity problem search, using the modules in `src/`.

## 1. Setup

```python
from pathlib import Path
import sys

import pandas as pd

PROJECT_DIR = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(PROJECT_DIR / "src"))

from recommend_problems import (
    load_problem_bank,
    recommend_problems,
    recommend_next_problem,
    format_recommendations,
)
from problem_similarity import similar_problems, format_similar_problems

DATA_FILE = PROJECT_DIR / "data" / "cleaned" / "codeforces_cleaned.csv"
RESULTS_DIR = PROJECT_DIR / "results"
RESULTS_DIR.mkdir(exist_ok=True)

pd.set_option("display.max_colwidth", 80)
```

## 2. Load the Problem Bank

```python
problem_bank = load_problem_bank(DATA_FILE)

print("Problems available:", len(problem_bank))
print("Rating range:", problem_bank["rating"].min(), "-", problem_bank["rating"].max())

problem_bank[["name", "rating", "tags"]].head()
```

## 3. Recommend Practice Problems by Target Rating

Recommend problems near a target rating, optionally filtered toward weak
tags the learner wants to practice.

```python
target_rating = 1800
weak_tags = ["dp", "greedy"]

recommendations = recommend_problems(
    target_rating=target_rating,
    weak_tags=weak_tags,
    count=8,
)

print(f"Practice problems near {target_rating} (weak tags: {weak_tags})\n")
print(format_recommendations(recommendations))

recommendations
```

## 4. Recommend Without Tag Filtering

```python
recommendations_no_tags = recommend_problems(
    target_rating=1400,
    weak_tags=None,
    count=8,
)

recommendations_no_tags
```

## 5. Similar Problems (TF-IDF + Cosine Similarity)

Given a problem name, URL, or partial name, find the most similar problems
in the dataset based on statement text, name, and tags.

```python
query_problem = "Odd-Even Subsequence"

similar = similar_problems(query_problem, count=8)

print(f"Problems similar to '{query_problem}'\n")
print(format_similar_problems(similar))

similar
```

## 6. Similarity Search With a Different Query

```python
query_problem_2 = problem_bank.iloc[50]["name"]

similar_2 = similar_problems(query_problem_2, count=5)

print(f"Problems similar to '{query_problem_2}'\n")
print(format_similar_problems(similar_2))
```

## 7. Next Problem Recommendation

After a learner solves a problem, recommend the next one: slightly harder,
and related by tags.

```python
solved_query = "Odd-Even Subsequence"

solved_problem, next_recommendations = recommend_next_problem(
    solved_problem_query=solved_query,
    count=5,
)

print(f"Solved: {solved_problem['name']} (rating {solved_problem['rating']})\n")
print("Suggested next problems:\n")
print(format_recommendations(next_recommendations))

next_recommendations
```

## 8. Save a Sample Recommendation Output

```python
sample_output = recommend_problems(target_rating=1600, weak_tags=["graphs"], count=10)
sample_output.to_csv(RESULTS_DIR / "sample_recommendations.csv", index=False)

print("Saved sample recommendations to", RESULTS_DIR / "sample_recommendations.csv")
sample_output
```
