```python
import pandas as pd

df = pd.read_csv("../data/cleaned/codeforces_cleaned.csv")

df.head()
df.shape
df.info()
df.isnull().sum()
df["rating"].describe()
df["rating"].value_counts().sort_index()


## Rating Distribution Analysis

import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

plt.hist(
    df["rating"],
    bins=20,
    edgecolor="black"
)

plt.title("Distribution of Codeforces Problem Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Problems")


plt.savefig(
    "../plots/rating_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

rating_counts = (
    df["rating"]
    .value_counts()
    .sort_index()
)

print(rating_counts)


print("Minimum Rating:", df["rating"].min())
print("Maximum Rating:", df["rating"].max())
print("Mean Rating:", df["rating"].mean())
print("Median Rating:", df["rating"].median())
```
